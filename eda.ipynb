```python
fraud_count = df['Class'].value_counts()[1]
legit_count = df['Class'].value_counts()[0]
fraud_pct = (fraud_count / len(df)) * 100

print(f"Fraudulent transactions: {fraud_count} ({fraud_pct:.3f}%)")
print(f"Legitimate transactions: {legit_count} ({100-fraud_pct:.3f}%)")
```

The dataset is highly imbalanced — fraud accounts for only ~0.17% of transactions. This means accuracy is not a reliable metric, and we'll need techniques like SMOTE and metrics like Precision-Recall AUC later in the project.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
sc = StandardScaler()
```

```python
df['scaled_amount'] = sc.fit_transform(df[['Amount']])
df['scaled_time'] = sc.fit_transform(df[['Time']])
```

```python
df.drop(['Amount','Time'], axis = 1, inplace = True)
```

```python
df.head()
```

```python
x = df.drop('Class',axis=1)
y = df['Class']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.2, random_state = 42 , stratify = y)
```

```python
np.save('X_train.npy', x_train)
np.save('X_test.npy', x_test)
np.save('y_train.npy', y_train)
np.save('y_test.npy', y_test)
```
